**Hypothesis Analysis**

**TTest** 

**Is there a significant difference in average Travel Cost between Male and Female customers?**

- **H0: There is no significant difference in average Travel Cost On Male and Female**

- **H1: There is a significant difference in average Travel Cost on Male and Female**

```python
MaleCustomer = Travel[Travel['Gender']=='Male']['Total_Trip_Cost']
FemaleCustomer = Travel[Travel['Gender']=='Female']['Total_Trip_Cost']
t_stat, p_value = st.ttest_ind(MaleCustomer, FemaleCustomer, equal_var = False)
print('T-Test: Travel Cost Difference Between Male and Female Customers:\n')
print(f"T-statistic: {t_stat:.2f}, P-value: {p_value:.2f}")
alpha = 0.05
if p_value < alpha:
    print("Reject H0")
    print("Average Cost is significantly different.")
else:
    print("Fail to reject H0")
    print("There is not enough evidence of a significant difference.")
```

Here are the meaningful insights from the hypothesis test:

- The test was conducted to check whether the average travel cost differs between male and female customers.
- Null hypothesis (H0): There is no significant difference in average travel cost between male and female customers.
- Alternative hypothesis (H1): There is a significant difference.

Key result:
- T-statistic = -0.92
- P-value = 0.36
- Significance level = 0.05

Interpretation:
- Since p = 0.36 > 0.05, we fail to reject H0.
- This means there is not enough statistical evidence to say that male and female customers spend significantly different amounts on travel.
- The observed difference in average travel cost is likely due to random variation rather than a real gender-based difference.

Business insight:
- Gender alone does not appear to be a meaningful factor in determining travel cost for this dataset.
- Pricing, promotions, or travel offers may not need to be adjusted separately for male and female customers based on this metric alone.
- The analysis suggests that other factors like destination, trip duration, booking class, season, or customer segment may have a stronger influence on travel cost.

Overall conclusion:
- The average travel cost is statistically similar across male and female customers in this dataset.
- The result indicates no strong evidence of a gender-based difference in travel spending.
